```python
import os
import pickle
from pathlib import Path

import pandas as pd
import matplotlib.pyplot as plt
from tqdm import tqdm

path = Path('/data/ljk/BatteryLife/datasets/processed/XJTU')
if not path.exists():
    raise FileNotFoundError(f'Path not found: {path}')

files = sorted([i for i in os.listdir(path) if i.endswith('.pkl')])
print(f'total files: {len(files)}')

MAX_CYCLES_PER_FILE = 21  # plot first 21 cycles (0..20)

ok = 0
failed = 0

for file in tqdm(files, desc='Plotting XJTU files'):
    try:
        with open(path / file, 'rb') as f:
            cell_data = pickle.load(f)

        filename = file.split('.pkl')[0]
        cycle_data_all = cell_data.get('cycle_data', [])
        length = len(cycle_data_all)

        currents = []
        voltages = []
        times = []

        time_offset = 0.0
        for i in range(min(length, MAX_CYCLES_PER_FILE)):
            cycle_data = cycle_data_all[i]

            cycle_df = pd.DataFrame({
                'current': cycle_data.get('current_in_A', []),
                'voltage': cycle_data.get('voltage_in_V', []),
                'charge_capacity': cycle_data.get('charge_capacity_in_Ah', []),
                'discharge_capacity': cycle_data.get('discharge_capacity_in_Ah', []),
                'test_time': cycle_data.get('time_in_s', []),
            })

            if cycle_df.empty:
                continue

            # Drop rows with invalid numbers to avoid plotting errors.
            cycle_df = cycle_df.apply(pd.to_numeric, errors='coerce').dropna(subset=['current', 'voltage', 'test_time'])
            if cycle_df.empty:
                continue

            t = cycle_df['test_time'].to_numpy(dtype=float)
            t = t - t[0] + time_offset
            time_offset = float(t[-1])

            currents.extend(cycle_df['current'].tolist())
            voltages.extend(cycle_df['voltage'].tolist())
            times.extend(t.tolist())

        if len(times) == 0:
            failed += 1
            print(f'[SKIP] {filename}: no valid points to plot')
            continue

        fig, ax1 = plt.subplots(figsize=(10, 6))
        ax1.plot(times, voltages, 'b-', linewidth=1.0)
        ax1.set_xlabel('Time(s)')
        ax1.set_ylabel('Voltage', color='b')
        ax1.tick_params(axis='y', colors='b')

        ax2 = ax1.twinx()
        ax2.plot(times, currents, 'r-', linewidth=1.0)
        ax2.set_ylabel('Current', color='r')
        ax2.tick_params(axis='y', colors='r')

        plt.title(f'Voltage-Current vs Time Profile\nFile: {filename}')
        plt.tight_layout()
        plt.show()
        plt.close(fig)

        ok += 1

    except Exception as e:
        failed += 1
        print(f'[FAIL] {file}: {type(e).__name__}: {e}')

print('\nDone')
print(f'ok={ok}')
print(f'failed={failed}')
```
