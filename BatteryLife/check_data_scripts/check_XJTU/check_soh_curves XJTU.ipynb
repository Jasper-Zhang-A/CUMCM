```python
import os
import pickle
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from scipy.interpolate import PchipInterpolator
from tqdm import tqdm

path = Path('/data/ljk/BatteryLife/datasets/processed/XJTU-new/')
files = sorted([f for f in os.listdir(path) if f.endswith('.pkl')])

if not files:
    raise FileNotFoundError(f'No .pkl files found in {path}')

print(f'Total files: {len(files)}')


def get_cycle_raw_discharge_capacity(cycle_data):
    cycle_df = pd.DataFrame({
        'current': cycle_data['current_in_A'],
        'discharge_capacity': cycle_data['discharge_capacity_in_Ah'],
    })
    discharge_seg = cycle_df.loc[cycle_df['current'] < 0, 'discharge_capacity']
    if not discharge_seg.empty:
        return float(discharge_seg.max())
    return float(cycle_df['discharge_capacity'].max())


def is_rpt_cycle(cycle_data):
    attribute = str(cycle_data.get('attribute', 'Cycling'))
    if attribute == 'RPT':
        return True
    cycle_desc = str(cycle_data.get('step_description', '')).lower()
    return 'test capacity' in cycle_desc or 'capacity test' in cycle_desc


def estimate_full_discharge_capacities(cycle_list):
    raw_capacities = np.asarray([
        get_cycle_raw_discharge_capacity(cycle_data)
        for cycle_data in cycle_list
    ], dtype=float)
    cycles = np.arange(1, len(raw_capacities) + 1, dtype=float)
    rpt_mask = np.asarray([is_rpt_cycle(cycle_data) for cycle_data in cycle_list], dtype=bool)
    valid = rpt_mask & np.isfinite(raw_capacities)

    if np.count_nonzero(valid) < 2:
        return raw_capacities, rpt_mask

    interpolator = PchipInterpolator(
        cycles[valid],
        raw_capacities[valid],
        extrapolate=True,
    )
    estimated = np.asarray(interpolator(cycles), dtype=float)
    estimated[~np.isfinite(estimated)] = raw_capacities[~np.isfinite(estimated)]
    return estimated, rpt_mask


for file_name in tqdm(files):
    file_path = path / file_name
    with open(file_path, 'rb') as f:
        cell_data = pickle.load(f)

    filename = file_name.replace('.pkl', '')
    cycle_list = cell_data['cycle_data']
    nominal_capacity = float(cell_data['nominal_capacity_in_Ah'])
    soc_interval = cell_data.get('SOC_interval', [0, 1])

    if isinstance(soc_interval, (list, tuple)) and len(soc_interval) >= 2:
        soc_scale = soc_interval[1] - soc_interval[0]
    else:
        soc_scale = 1.0

    if soc_scale == 0:
        soc_scale = 1.0

    cycles = np.arange(1, len(cycle_list) + 1, dtype=int)
    comparable_capacities, rpt_mask = estimate_full_discharge_capacities(cycle_list)
    raw_capacities = np.asarray([
        get_cycle_raw_discharge_capacity(cycle_data)
        for cycle_data in cycle_list
    ], dtype=float)
    raw_soh = comparable_capacities / nominal_capacity / soc_scale
    test_cycle_numbers = cycles[rpt_mask].tolist()
    test_cycle_indices = list(range(1, len(test_cycle_numbers) + 1))
    test_soh = raw_soh[rpt_mask].tolist()

    low_idx = np.where(raw_soh <= 0.1)[0] + 1

    print(f'\n{filename}')
    print(f'  Cycles: {len(cycles)}, RPT cycles: {len(test_cycle_numbers)}')
    if len(test_cycle_numbers) > 0:
        print(f'  First RPT cycle number: {test_cycle_numbers[0]}, plotted RPT index starts at {test_cycle_indices[0]}')
    if len(low_idx) > 1:
        gap = np.diff(low_idx)
        print(f'  Low SOH cycles (<=0.1) count={len(low_idx)}, gap median={float(np.median(gap)):.1f}, unique gaps={sorted(set(gap.tolist()))}')
    elif len(low_idx) == 1:
        print(f'  Low SOH cycles (<=0.1): {low_idx.tolist()}')

    if test_soh:
        fig, axes = plt.subplots(1, 2, figsize=(14, 4))
    else:
        fig, axes = plt.subplots(1, 1, figsize=(11, 4))
        axes = [axes]

    axes[0].plot(cycles, raw_capacities / nominal_capacity / soc_scale, linewidth=1.0, alpha=0.35, label='Raw per-cycle SOH')
    axes[0].plot(cycles, raw_soh, linewidth=1.4, alpha=0.85, label='Notebook PCHIP comparable SOH')
    axes[0].set_xlabel('Cycle number')
    axes[0].set_ylabel('SOH')
    axes[0].set_title(filename)
    axes[0].grid(alpha=0.3)
    axes[0].legend(loc='best')

    if test_soh:
        axes[1].plot(test_cycle_indices, test_soh, color='tab:orange', linewidth=2.0, marker='o', markersize=3, label='RPT SOH')
        axes[1].set_xlabel('RPT index')
        axes[1].set_ylabel('SOH')
        axes[1].set_title(f'{filename} RPT')
        axes[1].grid(alpha=0.3)
        axes[1].legend(loc='best')

    plt.tight_layout()
    plt.show()
```

```python
import pickle
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from scipy.interpolate import PchipInterpolator
from scipy.io import loadmat
from tqdm import tqdm

raw_root_candidates = [
    Path('/data/ljk/BatteryLife/datasets/raw/XJTU/Battery Dataset'),
    Path('/data/ljk/BatteryLife/datasets/raw/XJTU'),
]
pkl_root = Path('/data/ljk/BatteryLife/datasets/processed/XJTU')

raw_root = None
for candidate in raw_root_candidates:
    if candidate.exists():
        raw_root = candidate
        break
if raw_root is None:
    raise FileNotFoundError(f'Cannot find raw XJTU folder in: {raw_root_candidates}')

# XJTU preprocessing removes cycle 1 for these batches, so raw side must do the same before comparing.
batches_remove_first_cycle = {'Batch-1', 'Batch-2', 'Batch-3', 'Batch-4', 'Batch-6'}

pkl_files = sorted([p for p in pkl_root.glob('*.pkl')])
if not pkl_files:
    raise FileNotFoundError(f'No .pkl files found in {pkl_root}')

raw_mat_paths = sorted(raw_root.glob('Batch-*/*.mat'))
if not raw_mat_paths:
    raise FileNotFoundError(f'No .mat files found under {raw_root}')

# Build an index by raw stem (e.g., R3_battery-4) -> [(batch, path), ...]
raw_index = {}
for mat_path in raw_mat_paths:
    key = mat_path.stem
    raw_index.setdefault(key, []).append((mat_path.parent.name, mat_path))

print(f'Raw root: {raw_root}')
print(f'Total pkl files: {len(pkl_files)}')
print(f'Total raw mat files: {len(raw_mat_paths)}')

def mat_get_value(data, cycle, variable):
    var_names = [
        'system_time', 'relative_time_min', 'voltage_V', 'current_A',
        'capacity_Ah', 'power_Wh', 'temperature_C', 'description'
    ]
    if isinstance(variable, str):
        variable = var_names.index(variable)
    value = data[0][cycle - 1][variable]
    if variable == 7:
        return value[0]
    return value.reshape(-1)

def extract_raw_capacity_curve(mat_path, batch_name):
    mat = loadmat(str(mat_path))
    data = mat['data']
    cycle_count = data.shape[1]

    cycles = []
    capacities = []
    for cycle in range(1, cycle_count + 1):
        current = np.asarray(mat_get_value(data, cycle, 'current_A'), dtype=float)
        capacity = np.asarray(mat_get_value(data, cycle, 'capacity_Ah'), dtype=float)

        valid = np.isfinite(current) & np.isfinite(capacity)
        current = current[valid]
        capacity = capacity[valid]

        if capacity.size == 0:
            qd = np.nan
        else:
            discharge_mask = current <= -0.01
            if np.any(discharge_mask):
                qd = float(np.nanmax(capacity[discharge_mask]))
            else:
                qd = float(np.nanmax(capacity))

        cycles.append(cycle)
        capacities.append(qd)

    if batch_name in batches_remove_first_cycle and len(cycles) > 0:
        cycles = cycles[1:]
        capacities = capacities[1:]

    return np.asarray(cycles, dtype=int), np.asarray(capacities, dtype=float)

def extract_pkl_raw_capacity_curve(pkl_path):
    with open(pkl_path, 'rb') as f:
        cell_data = pickle.load(f)

    cycle_list = cell_data['cycle_data']
    capacities = []

    for cycle_data in cycle_list:
        current = np.asarray(cycle_data['current_in_A'], dtype=float)
        discharge_capacity = np.asarray(cycle_data['discharge_capacity_in_Ah'], dtype=float)

        valid = np.isfinite(current) & np.isfinite(discharge_capacity)
        current = current[valid]
        discharge_capacity = discharge_capacity[valid]

        if discharge_capacity.size == 0:
            qd = np.nan
        else:
            discharge_mask = current < 0
            if np.any(discharge_mask):
                qd = float(np.nanmax(discharge_capacity[discharge_mask]))
            else:
                qd = float(np.nanmax(discharge_capacity))

        capacities.append(qd)

    cycles = np.arange(1, len(capacities) + 1, dtype=int)
    return cycles, np.asarray(capacities, dtype=float)

def is_rpt_cycle(cycle_data):
    attribute = str(cycle_data.get('attribute', 'Cycling'))
    if attribute == 'RPT':
        return True
    cycle_desc = str(cycle_data.get('step_description', '')).lower()
    return 'test capacity' in cycle_desc or 'capacity test' in cycle_desc


def estimate_notebook_full_discharge_capacity_curve(pkl_path):
    """Comparable full-discharge curve estimated in this notebook, not stored in pkl."""
    with open(pkl_path, 'rb') as f:
        cell_data = pickle.load(f)

    cycle_list = cell_data['cycle_data']
    capacities = []
    rpt_mask = []

    for cycle_data in cycle_list:
        current = np.asarray(cycle_data['current_in_A'], dtype=float)
        discharge_capacity = np.asarray(cycle_data['discharge_capacity_in_Ah'], dtype=float)
        valid = np.isfinite(current) & np.isfinite(discharge_capacity)
        current = current[valid]
        discharge_capacity = discharge_capacity[valid]
        if discharge_capacity.size == 0:
            capacities.append(np.nan)
        else:
            discharge_mask = current < 0
            vals = discharge_capacity[discharge_mask] if np.any(discharge_mask) else discharge_capacity
            capacities.append(float(np.nanmax(vals)))
        rpt_mask.append(is_rpt_cycle(cycle_data))

    cycles = np.arange(1, len(capacities) + 1, dtype=int)
    capacities = np.asarray(capacities, dtype=float)
    valid = np.asarray(rpt_mask, dtype=bool) & np.isfinite(capacities)
    if np.count_nonzero(valid) < 2:
        return cycles, capacities

    interpolator = PchipInterpolator(
        cycles[valid].astype(float),
        capacities[valid],
        extrapolate=True,
    )
    estimated = np.asarray(interpolator(cycles.astype(float)), dtype=float)
    estimated[~np.isfinite(estimated)] = capacities[~np.isfinite(estimated)]
    return cycles, estimated

for pkl_path in tqdm(pkl_files, desc='Compare raw(.mat) vs processed(.pkl/life-label curve)'):
    pkl_name = pkl_path.stem
    raw_key = pkl_name.replace('XJTU_', '', 1)

    matches = raw_index.get(raw_key, [])
    if len(matches) == 0:
        print(f'[Skip] No raw .mat match for {pkl_name}')
        continue
    if len(matches) > 1:
        print(f'[Skip] Multiple raw .mat matches for {pkl_name}: {[str(m[1]) for m in matches]}')
        continue

    batch_name, mat_path = matches[0]

    raw_cycles, raw_caps = extract_raw_capacity_curve(mat_path, batch_name)
    pkl_cycles, pkl_caps = extract_pkl_raw_capacity_curve(pkl_path)
    life_cycles, life_caps = estimate_notebook_full_discharge_capacity_curve(pkl_path)

    n = min(len(raw_caps), len(pkl_caps), len(life_caps))
    if n == 0:
        print(f'[Skip] Empty capacity curve: {pkl_name}')
        continue

    cmp_df = pd.DataFrame({
        'cycle': np.arange(1, n + 1, dtype=int),
        'capacity_raw_mat_Ah': raw_caps[:n],
        'capacity_processed_pkl_raw_Ah': pkl_caps[:n],
        'capacity_notebook_pchip_full_Ah': life_caps[:n],
    })
    cmp_df['delta_pkl_raw_minus_mat_Ah'] = cmp_df['capacity_processed_pkl_raw_Ah'] - cmp_df['capacity_raw_mat_Ah']
    cmp_df['delta_notebook_pchip_minus_mat_Ah'] = cmp_df['capacity_notebook_pchip_full_Ah'] - cmp_df['capacity_raw_mat_Ah']

    print(f'\n{pkl_name} | {batch_name}')
    print(f'  raw cycles={len(raw_caps)}, pkl raw cycles={len(pkl_caps)}, notebook PCHIP curve cycles={len(life_caps)}, compared={n}')
    print(cmp_df.to_string(index=False))

    fig, axes = plt.subplots(1, 2, figsize=(14, 4.2))

    # Left panel: show full lengths for each source instead of truncating to shortest.
    axes[0].plot(raw_cycles, raw_caps, label='Raw (.mat)', linewidth=1.4)
    axes[0].plot(pkl_cycles, pkl_caps, label='Processed raw curve (.pkl)', linewidth=1.4, alpha=0.85)
    axes[0].plot(life_cycles, life_caps, label='Notebook PCHIP full-capacity curve', linewidth=1.8, alpha=0.9)
    axes[0].set_xlabel('Cycle number')
    axes[0].set_ylabel('Capacity (Ah)')
    axes[0].set_title(f'{pkl_name} capacity-cycle')
    axes[0].grid(alpha=0.3)
    axes[0].legend(loc='best')

    # Right panel remains aligned by shortest common length for fair delta comparison.
    axes[1].plot(cmp_df['cycle'], cmp_df['delta_pkl_raw_minus_mat_Ah'], color='tab:red', linewidth=1.2, label='Delta raw pkl - mat')
    axes[1].plot(cmp_df['cycle'], cmp_df['delta_notebook_pchip_minus_mat_Ah'], color='tab:green', linewidth=1.2, label='Delta notebook PCHIP - mat')
    axes[1].axhline(0.0, color='k', linestyle='--', linewidth=1.0, alpha=0.7)
    axes[1].set_xlabel('Cycle number')
    axes[1].set_ylabel('Delta capacity (Ah)')
    axes[1].set_title('Per-cycle difference')
    axes[1].grid(alpha=0.3)
    axes[1].legend(loc='best')

    plt.tight_layout()
    plt.show()

```
